```python
import pandas as pd 
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.model_selection import cross_val_score
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
```

```python
data=pd.read_excel("irisdata.xlsx")
split=StratifiedShuffleSplit(n_splits=1,test_size=0.2,random_state=42)
# print(data.info())
for train_index,test_index in split.split(data,data["Species"]):
    train_set=data.loc[train_index]
    test_set=data.loc[test_index]
training_set=train_set.iloc[:,:-1]
ytrain=train_set.iloc[:,-1]
testing_set=test_set.iloc[:,:-1]
ytest=test_set.iloc[:,-1]
```

## Training model

```python
model1=RandomForestClassifier()
model1.fit(training_set,ytrain)
a1=cross_val_score(model1,training_set,ytrain,cv=3,scoring="accuracy")
a1m=a1.mean()
```

```python
predictions1=model1.predict(testing_set)
```

## Checking accuracy of our model

```python
count=0
n=len(ytest)
for i in range(0,n):
    if ytest.iloc[i]==predictions1[i]:
        count+=1
acc1=(count*100)/n

print(f"Cross Validation scores for Decision Tree classifier is as follows:")
print(f"{a1}")
print(a1m)
print(f"Accuracy of Random Forest Classifier = {acc1}%")
```

## Using Gaussian NB Model

```python
model2=GaussianNB()
model2.fit(training_set,ytrain)
a2=cross_val_score(model2,training_set,ytrain,cv=3,scoring="accuracy")
a2m=a2.mean()
predictions2=model2.predict(testing_set)
print(f"Cross Validation scores for Decision Tree classifier is as follows:")
print(f"{a2}")
print(a2m)
count=0
n=len(ytest)
for i in range(0,n):
    if ytest.iloc[i]==predictions2[i]:
        count+=1
acc2=(count*100)/n
print(f"Accuracy of Gaussian NB = {acc2}%")
```

## Decision Tree Classifier

```python
model3=DecisionTreeClassifier()
model3.fit(training_set,ytrain)
a3=cross_val_score(model3,training_set,ytrain,cv=3,scoring="accuracy")
a3m=a3.mean()
predictions3=model3.predict(testing_set)
print(f"Cross Validation scores for Decision Tree classifier is as follows:")
print(f"{a3}")
print(a3m)
count=0
n=len(ytest)
for i in range(0,n):
    if ytest.iloc[i]==predictions3[i]:
        count+=1
acc3=(count*100)/n
print(f"Accuracy of Decision Tree Classifier = {acc3}%")

```
